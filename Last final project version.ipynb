# 3- Feature Engineering

```python
# 1. Total annual income 
df["Total_Income"] = df["Annual_Income"] + (df["Other_Income"] * 12)

# 2. Loan-to-income ratio
df["Loan_to_Income"] = df["Loan_Amount"] / df["Annual_Income"]

# 3. EMI-to-income ratio
df["EMI_to_Income"] = df["Monthly_EMI"] / df["Monthly_Income"]

# 4. Experience-to-age ratio
df["Experience_Ratio"] = df["Total_Work_Experience"] / df["Age"]

# 5. Loan burden ratio  // الالتزمات مقارنة بالدخل السنوي
df["Loan_Burden"] = (
    df["Existing_Loan_Amount"] + df["Loan_Amount"]
) / df["Annual_Income"]
```

# 6 - Data Preprocessing

```python
# Drop `Loan_Status_num` column due to high correlation with Target column
df.drop('Loan_Status_num', axis = 1, inplace= True)
```

###  1- Split Data into Input & Output Features

```python
x = df.drop(['Loan_Status'], axis=1)
y = df['Loan_Status']
```

### 2- Split Data Into Train & Test

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42, stratify=y)
```

```python
y_train.value_counts(normalize= True) * 100
```

```python
y_test.value_counts(normalize= True) * 100
```

## 6.2 - Numerical Handling

```python
numeric_features = x.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = x.select_dtypes(
    include=["object"]
).columns
```

```python
from sklearn import set_config

set_config(display="diagram")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
num_pipeline = Pipeline([('scaler', RobustScaler())])
num_pipeline
```

### 6.3 - Categorical Handling

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    ))
])
categorical_pipeline
```

#### 6.4 - ColumnTransformer

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", num_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

preprocessor
```

# 7 - Machine Learning

### 7.1 - Check All Models without changing in hyperparameters

```python
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier


models = [
    ("Logistic Regression", LogisticRegression(max_iter=1000)),
    ("Decision Tree", DecisionTreeClassifier(random_state=42)),
    ("Random Forest", RandomForestClassifier(random_state=42))
]


for name, model in models:

    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipe,
        x,
        y,
        cv=5,
        scoring="accuracy",
        return_train_score=True,
        n_jobs=-1
    )

    train_accuracy = scores["train_score"].mean()
    test_accuracy = scores["test_score"].mean()

    print("=" * 40)
    print(name)
    print(f"Train Accuracy: {train_accuracy * 100:.2f}%")
    print(f"Test Accuracy : {test_accuracy * 100:.2f}%")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

models = [
    ("Logistic Regression", LogisticRegression(max_iter=1000)),
    ("Decision Tree", DecisionTreeClassifier(random_state=42)),
    ("Random Forest", RandomForestClassifier(random_state=42))
]

for name, model in models:

    model_pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("Model", model)
    ])

    # Training
    model_pipeline.fit(x_train, y_train)

    # Prediction
    y_pred = model_pipeline.predict(x_test)

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, pos_label= 1)
    recall = recall_score(y_test, y_pred, pos_label= 1)
    f1 = f1_score(y_test, y_pred, pos_label= 1)

    print(name)
    print("-" * 40)
    print("Accuracy :", round(accuracy * 100, 2), "%")
    print("Precision:", round(precision * 100, 2), "%")
    print("Recall   :", round(recall * 100, 2), "%")
    print("F1 Score :", round(f1 * 100, 2), "%")

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    print("=" * 50)
```

```python
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("Model", RandomForestClassifier(random_state=42))
])
```

### 7.2 - Check All Models with changing in hyperparameters (Gridsearch)

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "Model__n_estimators": [100, 200, 300],
    "Model__max_depth": [10, 20, 30],
    "Model__min_samples_split": [2, 5],
    "Model__min_samples_leaf": [1, 2]
}

grid = GridSearchCV(
    rf_pipeline,
    param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

grid.fit(x_train, y_train)

print("Best Parameters:")
print(grid.best_params_)

print("Best F1:", grid.best_score_)
```

```python
best_model = grid.best_estimator_

y_pred = best_model.predict(x_test)
```

```python
print("Tuned Random Forest")
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, pos_label=1))
print("Recall   :", recall_score(y_test, y_pred, pos_label=1))
print("F1 Score :", f1_score(y_test, y_pred, pos_label=1))
```
